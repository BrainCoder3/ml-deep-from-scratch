```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def build_audit(path):
    df = pd.read_csv(path)
    length = len(df)
    typ = type(df)
    num_nan = df.isna().sum()
    num_duplicated = df.duplicated().sum()
    return df,length, typ, num_nan, num_duplicated


def build_analys(dataset):
    display(dataset.describe())
    return True


def ploting_build(dataset,A,B):
    plt.figure(figsize=(12,8))
    plt.subplot(1,2,1)
    plt.hist(dataset[A])
    plt.subplot(1,2,2)
    plt.plot(dataset[B])
    plt.show()

def pipeline_build(path,A,B):
    df, length, typ, num_nan, num_duplicated = build_audit(path)
    print(f"Length : {length} | type : {typ} | num_nan : {num_nan} | num_duplicated : {num_duplicated}")
    result = build_analys(df)
    ploting_build(df,A,B)

```
